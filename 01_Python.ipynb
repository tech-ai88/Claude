# 연산자
- 산술 연산자, 비교(관계) 연산자, 논리 연산자, 대입(할당) 연산자

## 산술 연산자

```python
a=10
b=3
print(a + b)
print(a - b)
print(a * b)
print(a / b)    # 나누기(몫) : 소수점까지 구합
print(a // b)   # 나누기(몫) : 정수
print(a % b)    # 나머지
print(a ** b)   # 거듭제곱
```

## 관계(비교) 연산자

```python
x=10
y=20
print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

## 논리 연산자

```python
su=10
value=20
result1=su > 0 and value < 20
result2=su == 0 or value == 20
result3=not(su % 2 == 0)

print(result1)
print(result2)
print(result3)
```

## 대입 연산자

```python
num=100
num +=2             # 102
num -=2             # 100
num *=2             # 200 
num //=2            # 100
print(num)
```

# 제어문
## 조건문
- if문, if ~ else문, if ~ elif문

```python
su=20

if su > 10:     # 조건식
    print("입력한 수는 10보다 큰 수 입니다.")   # 문장 또는 블록
```

```python
if su % 2 == 0:
    print("짝수")
else:
    print("홀수")
```

```python
jumsu=78

if jumsu >= 90:
    print("A 등급")
elif jumsu >= 80:
    print("B 등급")
elif jumsu >= 70:
    print("C 등급")
else:
    print("D 등급")
```

## 반복문

- 일정 수치 반복

```python
for i in range(0, 8, 2):
    print(i, end=' ')
```

```python
for i in range(2, 8):
    print(i, end=' ')
```

```python
for i in range(6, 0, -1):
    print(i, end=' ')
```

```python
for i in range(5):
    print(i, end=' ')
```

- 문자 반복

```python
for i in range(10):
    print("안녕하세요")
```

- 계산 반복

```python
hap=0

for i in range(1, 11):
    hap += i
    print(i, hap)
```
