````python
# 클래스 이름이 "ai_effect_symbol"인 요소를 ai_s_else 변수에 할당
ai_s_else = driver.find_element(By.CLASS_NAME, "ai_effect_symbol")

# 버튼 클릭
ai_s_else.click()
````


````python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

# 대기시간
st = 3

# 검색할 여름 전자제품
keywords = [
    "에어컨",
    "선풍기",
    "제습기",
    "냉풍기",
    "서큘레이터"
]

# ==================================================
# 키워드별 반복
# ==================================================
for keyword in keywords:

    print("=" * 70)
    print(f"검색 키워드 : {keyword}")
    print("=" * 70)

    # 1) 웹 브라우저 열기
    driver = webdriver.Chrome()

    # 2) 네이버 접속
    driver.get("https://www.naver.com")
    time.sleep(st)

    # 3) 검색창 찾기
    green_box_ele = driver.find_element(By.ID, "query")

    # 4) 검색어 입력
    green_box_ele.send_keys(keyword)

    # 5) 검색 버튼 클릭
    ai_s_else = driver.find_element(By.CLASS_NAME, "ai_effect_symbol")
    ai_s_else.click()
    time.sleep(st)

    # 6) 카페 메뉴 클릭
    elements = driver.find_elements(By.CLASS_NAME, "wxxES_QvNoYHFNCZ")

    for element in elements:
        if element.text.strip() == "카페":
            element.click()
            print("카페 메뉴를 클릭했습니다.")
            break

    time.sleep(st)

    # 7) 게시글 제목 및 URL 출력
    title_links = driver.find_elements(By.CLASS_NAME, "title_link")

    for idx, element in enumerate(title_links, start=1):

        title = element.text.strip()
        url = element.get_attribute("href")

        print(f"[{idx}]")
        print("제목 :", title)
        print("URL  :", url)
        print("-" * 50)

    # 브라우저 종료
    driver.quit()

    print(f"{keyword} 검색 완료\n")
````

## 엑셀로 저장하기
- [프롬프트]
```
- 위의 코드를 기반으로 아래의 기능을 추가하여 발전시켜줘
    - 출력되는 내용을 기반으로 엑셀로 제작하기
    - 엑셀의 구조는 ["키워드", 게시글제목", "URL"]
    - 엑셀저장경로 C:\teacher\결과\웹
    - 엑셀파일명 "naver_cafe_elect01.xlsx"
```

````python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import pandas as pd
import os


# 대기시간
st = 3


# 검색할 여름 전자제품
keywords = [
    "에어컨",
    "선풍기",
    "제습기",
    "냉풍기",
    "서큘레이터"
]


# 엑셀 저장 리스트
result_list = []


# ==================================================
# 키워드 반복 검색
# ==================================================
for keyword in keywords:

    print("=" * 70)
    print(f"검색 키워드 : {keyword}")
    print("=" * 70)


    # 1) 크롬 실행
    driver = webdriver.Chrome()


    # 2) 네이버 접속
    driver.get("https://www.naver.com")
    time.sleep(st)


    # 3) 검색창 찾기
    green_box_ele = driver.find_element(
        By.ID,
        "query"
    )


    # 4) 검색어 입력
    green_box_ele.send_keys(keyword)


    # 5) 검색 버튼 클릭
    ai_s_else = driver.find_element(
        By.CLASS_NAME,
        "ai_effect_symbol"
    )

    ai_s_else.click()

    time.sleep(st)


    # 6) 카페 메뉴 클릭
    elements = driver.find_elements(
        By.CLASS_NAME,
        "YTTb5CeLN3ks4qzz"
    )


    for element in elements:

        if element.text.strip() == "카페":

            element.click()

            print("카페 메뉴 클릭 완료")

            break


    time.sleep(st)


    # 7) 게시글 제목 / URL 수집
    title_links = driver.find_elements(
        By.CLASS_NAME,
        "title_link"
    )


    for idx, element in enumerate(title_links, start=1):

        title = element.text.strip()

        url = element.get_attribute("href")


        print(f"[{idx}]")
        print("제목 :", title)
        print("URL :", url)
        print("-"*50)


        # ===============================
        # Excel 저장용 데이터 추가
        # ===============================
        result_list.append(
            {
                "키워드": keyword,
                "게시글제목": title,
                "URL": url
            }
        )


    # 브라우저 종료
    driver.quit()


    print(f"{keyword} 검색 완료\n")



# ==================================================
# DataFrame 생성
# ==================================================

df = pd.DataFrame(result_list)


print("\n수집 데이터 확인")
print(df)



# ==================================================
# Excel 저장
# ==================================================

save_path = r"C:\teacher\결과\웹"


# 폴더 없으면 생성
if not os.path.exists(save_path):
    os.makedirs(save_path)


file_name = "naver_cafe_elect.xlsx"


full_path = os.path.join(
    save_path,
    file_name
)


# Excel 저장
df.to_excel(
    full_path,
    index=False
)


print("="*70)
print("Excel 저장 완료")
print(full_path)
print("="*70)
````

```
- [프롬프트]
    - 지메일 자동으로 전송하는 코드 작성해줘
    - [보내는 사람] : jwaminhye@gmail.com
    - [받는사람] : jwaminhye@gmail.com
    - [파일경로] : C:\teacher\결과\웹 엑셀파일명 "naver_cafe_elect01.xlsx"
    - [제목] : 여름 냉방 상품 5개 추천
    - [내용] : 네이버에서 5개의 냉방 상품을 검색후, 카페 게시글의 갯수를 정리해서 보냅니다.
- 비밀번호 만든법 : PPT 참조) 
```

````python
import smtplib
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart
from email.mime.base import MIMEBase
from email import encoders
import os


# ===============================
# Gmail 정보
# ===============================

SENDER = "jwaminhye@gmail.com"
RECEIVER = "jwaminhye@gmail.com"

# Gmail 앱 비밀번호 입력 (16자리)
APP_PASSWORD = "본인 비밀번호"


# ===============================
# 메일 정보
# ===============================

SUBJECT = "여름 냉방 상품 5개 추천"

BODY = """
네이버에서 5개의 냉방 상품을 검색후,
카페 게시글의 갯수를 정리해서 보냅니다.
"""


# ===============================
# 첨부파일 경로
# ===============================

FILE_PATH = r"C:\teacher\결과\웹\naver_cafe_elect01.xlsx"


# ===============================
# 이메일 생성
# ===============================

msg = MIMEMultipart()

msg["From"] = SENDER
msg["To"] = RECEIVER
msg["Subject"] = SUBJECT


# 본문 추가
msg.attach(
    MIMEText(BODY, "plain", "utf-8")
)


# ===============================
# 파일 첨부
# ===============================

if os.path.exists(FILE_PATH):

    filename = os.path.basename(FILE_PATH)

    with open(FILE_PATH, "rb") as attachment:
        part = MIMEBase(
            "application",
            "octet-stream"
        )
        part.set_payload(attachment.read())

    encoders.encode_base64(part)

    part.add_header(
        "Content-Disposition",
        f"attachment; filename={filename}"
    )

    msg.attach(part)

else:
    print("파일을 찾을 수 없습니다.")
    raise FileNotFoundError(FILE_PATH)


# ===============================
# Gmail SMTP 전송
# ===============================

try:

    server = smtplib.SMTP(
        "smtp.gmail.com",
        587
    )

    server.starttls()

    server.login(
        SENDER,
        APP_PASSWORD
    )

    server.sendmail(
        SENDER,
        RECEIVER,
        msg.as_string()
    )

    server.quit()

    print("메일 전송 완료!")

except Exception as e:
    print("메일 전송 실패:", e)
````
