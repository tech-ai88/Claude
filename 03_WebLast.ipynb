```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import pandas as pd
import os


# 대기시간
st = 3


# 검색할 여름 전자제품
keywords = [
    "에어컨",
    "선풍기",
    "제습기",
    "냉풍기",
    "서큘레이터"
]


# 엑셀 저장 리스트
result_list = []


# ==================================================
# 키워드 반복 검색
# ==================================================
for keyword in keywords:

    print("=" * 70)
    print(f"검색 키워드 : {keyword}")
    print("=" * 70)


    # 1) 크롬 실행
    driver = webdriver.Chrome()


    # 2) 네이버 접속
    driver.get("https://www.naver.com")
    time.sleep(st)


    # 3) 검색창 찾기
    green_box_ele = driver.find_element(
        By.ID,
        "query"
    )


    # 4) 검색어 입력
    green_box_ele.send_keys(keyword)


    # 5) 검색 버튼 클릭
    ai_s_else = driver.find_element(
        By.CLASS_NAME,
        "ai_effect_symbol"
    )

    ai_s_else.click()

    time.sleep(st)


    # 6) 카페 메뉴 클릭
    elements = driver.find_elements(
        By.CLASS_NAME,
        "YTTb5CeLN3ks4qzz"
    )


    for element in elements:

        if element.text.strip() == "카페":

            element.click()

            print("카페 메뉴 클릭 완료")

            break


    time.sleep(st)


    # 7) 게시글 제목 / URL 수집
    title_links = driver.find_elements(
        By.CLASS_NAME,
        "title_link"
    )


    for idx, element in enumerate(title_links, start=1):

        title = element.text.strip()

        url = element.get_attribute("href")


        print(f"[{idx}]")
        print("제목 :", title)
        print("URL :", url)
        print("-"*50)


        # ===============================
        # Excel 저장용 데이터 추가
        # ===============================
        result_list.append(
            {
                "키워드": keyword,
                "게시글제목": title,
                "URL": url
            }
        )


    # 브라우저 종료
    driver.quit()


    print(f"{keyword} 검색 완료\n")



# ==================================================
# DataFrame 생성
# ==================================================

df = pd.DataFrame(result_list)


print("\n수집 데이터 확인")
print(df)



# ==================================================
# Excel 저장
# ==================================================

save_path = r"C:\teacher\결과\웹"


# 폴더 없으면 생성
if not os.path.exists(save_path):
    os.makedirs(save_path)


file_name = "naver_cafe_elect.xlsx"


full_path = os.path.join(
    save_path,
    file_name
)


# Excel 저장
df.to_excel(
    full_path,
    index=False
)


print("="*70)
print("Excel 저장 완료")
print(full_path)
print("="*70)
```
