### 파일100개 생성하기
- [프롬프트] : 명확한 지시, 참조 텍스트 제공, 복잡한 작업 분해, 체계적인 테스트
```
- 위의 코드를 기반하여 반복문으로 확장해서 총 100개의 엑셀 파일이 나오게 발전 시켜줘
    - "엑셀1.xlsx" ~ "엑셀100.xlsx" 
```

````python
from openpyxl import Workbook
import os

# 저장 폴더
folder_path = r"C:\teacher\결과\엑셀"

# 폴더가 없으면 생성
os.makedirs(folder_path, exist_ok=True)

# 100개 엑셀 파일 생성
for i in range(1, 101):

    # 파일명 생성
    file_name = f"엑셀{i}.xlsx"
    file_path = os.path.join(folder_path, file_name)

    # 엑셀 파일 생성
    wb = Workbook()
    ws = wb.active
    ws.title = "Sheet1"

    # 데이터 입력
    ws["A1"] = "파일번호"
    ws["B1"] = "이름"
    ws["C1"] = "점수"

    ws.append([i, f"학생{i}", 90 + (i % 10)])

    # 저장
    wb.save(file_path)

    print(f"{file_name} 생성 완료")

print("총 100개의 엑셀 파일 생성 완료!")
````

## 파일 이름 변경
### 파일1개 이름 변경
- [프롬프트] : 명확한 지시
```
- "C:\teacher\결과\엑셀" 폴더에
```

### 파일100개 이름 변경
- [프롬프트] 명확한 지시, 참조 텍스트 제공, 복잡한 작업 분해, 체계적인 테스트
```
- 위의 코드를 기반하여 반복문으로 확장해서, 총 100개의 파일을 변경하는 코드 만들어줘
- [BEFORE] "엑셀1.xlsx" ~ "엑셀100.xlsx"
- [AFTER] "엑셀AAA1.xlsx" ~ "엑셀AAA100.xlsx"
```

````python
import os

# 작업 폴더
folder_path = r"C:\teacher\결과\엑셀"

# 1~100 파일명 변경
for i in range(1, 101):

    # BEFORE 파일명
    before_file = f"엑셀{i}.xlsx"

    # AFTER 파일명 (3자리 숫자)
    after_file = f"엑셀{i:03d}.xlsx"

    # 전체 경로
    before_path = os.path.join(folder_path, before_file)
    after_path = os.path.join(folder_path, after_file)

    # 파일이 존재하면 변경
    if os.path.exists(before_path):
        os.rename(before_path, after_path)
        print(f"{before_file} → {after_file}")

    else:
        print(f"파일 없음 : {before_file}")

print("파일명 변경 완료!")
````

## 파일 대량 자동 분류
### 현업에 있는 파일 100개 제작하는 기능
- [프롬프트] New Chat
```

-  있을법한 파일 100개 랜덤으로 제작하는 파이썬 코드 만들어줘
```

````python
import os
import random
import csv
from datetime import datetime
from openpyxl import Workbook
from docx import Document
from reportlab.pdfgen import canvas
from pptx import Presentation

# 저장 폴더
SAVE_DIR = r"C:\teacher\결과\회사"

os.makedirs(SAVE_DIR, exist_ok=True)

subjects = [
    "매출보고서",
    "생산현황",
    "품질점검결과",
    "회의록",
    "프로젝트계획",
    "인사평가",
    "예산계획",
    "거래처관리",
    "반도체생산현황",
    "공정개선",
    "고객불만분석",
    "재고현황",
    "출하현황",
    "개발일정",
    "시장조사",
    "AI사업계획",
    "투자계획",
    "교육자료",
    "보안점검",
    "업무일지"
]

extensions = [
    "xlsx",
    "docx",
    "pdf",
    "pptx",
    "txt",
    "csv"
]

names = [
    "김민수","이지훈","박서준","최현우","정유진",
    "한지민","오세훈","김도윤","이수빈","강지훈"
]

departments = [
    "DX부문","MX사업부","반도체사업부","품질혁신팀",
    "AI센터","인사팀","재무팀","영업팀","개발팀"
]


def make_txt(path):
    with open(path, "w", encoding="utf-8") as f:
        f.write("삼성전자 내부 문서\n\n")
        for _ in range(10):
            f.write(f"{random.choice(subjects)} 관련 내용입니다.\n")


def make_csv(path):
    with open(path, "w", newline="", encoding="utf-8-sig") as f:
        writer = csv.writer(f)
        writer.writerow(["사번", "이름", "부서", "점수"])
        for i in range(20):
            writer.writerow([
                f"E{1000+i}",
                random.choice(names),
                random.choice(departments),
                random.randint(60,100)
            ])


def make_xlsx(path):
    wb = Workbook()
    ws = wb.active
    ws.title = "데이터"

    ws.append(["사번","이름","부서","매출"])
    for i in range(30):
        ws.append([
            f"S{1000+i}",
            random.choice(names),
            random.choice(departments),
            random.randint(1000000,50000000)
        ])

    wb.save(path)


def make_docx(path):
    doc = Document()
    doc.add_heading("삼성전자 업무 문서", 0)

    for _ in range(8):
        doc.add_paragraph(
            f"{random.choice(subjects)}에 대한 검토 결과를 작성합니다. "
            f"담당자는 {random.choice(names)}이며 "
            f"{random.choice(departments)}에서 작성하였습니다."
        )

    doc.save(path)


def make_pdf(path):
    c = canvas.Canvas(path)
    y = 800

    c.setFont("Helvetica", 12)

    for i in range(20):
        c.drawString(
            50,
            y,
            f"Samsung Electronics Report Line {i+1}"
        )
        y -= 30

    c.save()


def make_ppt(path):
    prs = Presentation()

    for i in range(5):
        layout = prs.slide_layouts[1]
        slide = prs.slides.add_slide(layout)

        slide.shapes.title.text = random.choice(subjects)
        slide.placeholders[1].text = (
            f"담당 : {random.choice(names)}\n"
            f"부서 : {random.choice(departments)}\n"
            "업무 진행 현황 및 향후 계획"
        )

    prs.save(path)


makers = {
    "txt": make_txt,
    "csv": make_csv,
    "xlsx": make_xlsx,
    "docx": make_docx,
    "pdf": make_pdf,
    "pptx": make_ppt
}

used = set()

while len(used) < 100:

    subject = random.choice(subjects)
    ext = random.choice(extensions)

    filename = (
        f"{random.randint(2024,2026)}"
        f"{random.randint(1,12):02d}_"
        f"{subject}_{random.randint(1,999)}.{ext}"
    )

    if filename in used:
        continue

    used.add(filename)

    path = os.path.join(SAVE_DIR, filename)

    makers[ext](path)

print(f"{len(used)}개의 랜덤 파일 생성 완료!")
print(SAVE_DIR)
````
